```python
import pandas as pd

data = pd.read_csv("Listers.csv", sep=";")

data.head()
```

```python
data["revenue"] = data["revenue"].astype(float)

revenue_by_user = data.groupby("user_id")["revenue"].sum()

avg_revenue_per_user = revenue_by_user.mean()

print("Средний доход на пользователя:", round(avg_revenue_per_user, 2))
```

```python
# Оставляем по одному возрасту на каждого пользователя
age_by_user = data.groupby("user_id")["age"].first()

# Считаем медиану возраста пользователей
median_age = age_by_user.median()

print("Медиана возраста пользователя:", median_age)
```
